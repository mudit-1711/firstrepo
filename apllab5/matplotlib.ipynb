```python
X=[1,2,3,4,5,6,7,6,5,4,3]
Y=[]
for i in X :
    Y.append(-2*i -9)
kwargs = {
    'linestyle': '-',
    'marker': 'o',
    'color': 'blue'
}

plt.plot(X, Y, **kwargs)
plt.title('Plot 1')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
