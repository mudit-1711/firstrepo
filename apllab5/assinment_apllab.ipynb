two dice are rolled and sum is evaluated, plot the bar chart for multiple die rolls. number of die rolls will be entered by the user 
and make sure the number entered by user is always greater than 500

```python
import random
import matplotlib.pyplot as plt
n = int(input("Enter number of rolls "))
while n <= 500:
    print("number greater than 500.")
    n = int(input("Enter number of rolls: "))
sums = []
for i in range(n):
    d1 = random.randint(1, 6)
    d2 = random.randint(1, 6)
    sums.append(d1 + d2)
freq = {}
for i in range(2, 13):
    freq[i] = sums.count(i)
plt.bar(freq.keys(), freq.values(),color='red')
plt.xlabel("Sum of Two Dice")
plt.ylabel("Frequency")
plt.title("Distribution of Sums for Multiple Dice Rolls")
plt.show()
```

write a program to report prime factorization of a number

```python
n = int(input("Enter a number: "))
print("Prime factors:", end=" ")
i = 2
while n>1:
    while n%i==0:
        print(i,end=" ")
        n=n//i
    i+=1
```
