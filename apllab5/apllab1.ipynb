```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
arr = np.random.randint(10, 100, size=(6, 6))
print(arr)
```

```python
extracted = arr[::2, ::2]
print(extracted)
```

```python
divisible = arr[(arr % 3 == 0) & (arr % 5 == 0)]
print(divisible)
```

```python
mean = arr.mean()
arr[arr < 30] = mean
print(arr)
```

```python
arr[:, [0, -1]] = arr[:, [-1, 0]]
print(arr)
```

```python
unique_arr = np.unique(arr)
print(unique_arr)
```
